```python
cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin',
       'BMI', 'DiabetesPedigreeFunction', 'Age']
```

```python
for col in cols:
    print(f"{col} where value is null: {df[col].isna().sum()}")    
    print(f"{col} where value is 0: {(df[col]== 0).sum()}")
```

```python
cols_replace = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
```

```python
df[cols_replace] = df[cols_replace].replace(0, np.nan)
```

```python
df[cols_replace] = df[cols_replace].fillna(df[cols_replace].median())
```
